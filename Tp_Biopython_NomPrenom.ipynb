```python
from Bio.Seq import Seq
dna = Seq("ATGGCCATTGTAATGGGCCGC")
print("Séquence originale :", dna)
print("Complément :", dna.complement())
print("Complément inverse :", dna.reverse_complement())
print("Transcription en ARNm :", dna.transcribe())
```

Réponse 3.1 :

Le complément inverse est important car il permet d'obtenir la séquence complémentaire dans le sens 5' → 3', ce qui est essentiel pour l'analyse des séquences d'ADN et la réplication.

```python
from Bio.Seq import Seq
dna = Seq("ATGGCCATTGTAATGGGCCGC")
proteine = dna.translate()
print("Protéine :", proteine)
print("Longueur de la protéine :", len(proteine), "acides aminés")
```


Réponse 3.2 :

La protéine obtenue contient 7 acides aminés.

La séquence d'ADN contient 7 codons hors codon stop.

```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction


def gc_content(sequence):
   """Calcule le pourcentage de G+C dans une séquence."""
   seq_str = str(sequence).upper()
   nb_gc = seq_str.count("G") + seq_str.count("C")
   return 100 * nb_gc / len(seq_str)


dna = Seq("ATGGCCATTGTAATGGGCCGC")
print("Séquence :", dna)
print("Contenu en GC :", round(gc_content(dna), 2), "%")




print("Avec gc_fraction :", round(gc_fraction(dna) * 100, 2), "%")
```

Réponse 3.3 :

Oui, les deux résultats sont identiques (57,14 %).

Parce que les deux méthodes calculent le pourcentage des bases G et C dans la séquence d'ADN.

```python
from Bio import SeqIO



sequences_longues = []
for record in SeqIO.parse("echantillons.fasta", "fasta"):
     if len(record.seq) > 80:
         sequences_longues.append(record)


print("Séquences > 80 pb :")
for record in sequences_longues:
     print(f" {record.id} : {len(record.seq)} pb")


SeqIO.write(sequences_longues, "sequences_longues.fasta", "fasta")
print("\nFichier sequences_longues.fasta écrit.")
```

Réponse 4.2 :

Deux séquences ont été conservées après le filtrage : seq1 et seq2, car leur longueur est supérieure à 80 pb.

Ce type de filtrage est utile en microbiologie pour éliminer les séquences trop courtes, partielles ou de mauvaise qualité.

```python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

for record in SeqIO.parse("echantillons.fasta", "fasta"):
    gc = gc_fraction(record.seq) * 100


    print(f"{record.id} : {len(record.seq)} pb, GC = {gc:.1f}%")
```
